```python
from transformers import pipeline
```

```python
prompt = "The future of Artificial Intelligence is"

models = {
    "BERT": "bert-base-uncased",
    "RoBERTa": "roberta-base",
    "BART": "facebook/bart-base"
}

for name, model in models.items():
    try:
        gen = pipeline("text-generation", model=model, framework="pt")
        output = gen(prompt, max_length=40)
        print(f"\n{name} Output:")
        print(output[0]["generated_text"])
    except Exception as e:
        print(f"\n{name} Error:")
        print(e)
```

```python
sentence = "The goal of Generative AI is to [MASK] new content."

for name, model in models.items():
    try:
        fm = pipeline("fill-mask", model=model, framework="pt")
        result = fm(sentence)
        print(f"\n{name} Predictions:")
        for r in result[:3]:
            print(r["token_str"], "-", round(r["score"],3))
    except Exception as e:
        print(f"\n{name} Error:")
        print(e)
```

```python
context = "Generative AI poses significant risks such as hallucinations, bias, and deepfakes."
question = "What are the risks?"

for name, model in models.items():
    try:
        qa = pipeline("question-answering", model=model, framework="pt")
        res = qa(question=question, context=context)
        print(f"\n{name} Answer:", res["answer"])
    except Exception as e:
        print(f"\n{name} Error:")
        print(e)
```

| Task | Model | Classification (Success/Failure/Partial) | Observation (What actually happened?) | Why did this happen? (Architectural Reason) |
|-----|------|-------------------------------------------|---------------------------------------|---------------------------------------------|
| Generation | BERT | Failure | Generated only dots and meaningless characters | BERT is encoder-only and not designed for autoregressive generation |
| Generation | RoBERTa | Failure | Did not extend the prompt | RoBERTa is encoder-only |
| Generation | BART | Partial Success | Generated continuation but text is incoherent | Encoder-decoder can generate, but bart-base is not fine-tuned for free text generation |
| Fill-Mask | BERT | Success | Predicted create, generate, produce | Trained with Masked Language Modeling (MLM) |
| Fill-Mask | RoBERTa | Failure | Error: no mask token found | RoBERTa expects `<mask>` instead of `[MASK]` |
| Fill-Mask | BART | Failure | Error: no mask token found | BART uses `<mask>` token and not `[MASK]` |
| QA | BERT | Partial Success | Returned fragment of correct answer | Base BERT not fine-tuned for QA |
| QA | RoBERTa | Partial Success | Returned correct risk phrase | Strong encoder but not QA fine-tuned |
| QA | BART | Partial Success | Returned correct phrase | Encoder-decoder handles seq2seq QA reasonably |

RoBERTa and BART require the `<mask>` token instead of `[MASK]`. Using the correct token resolves the fill-mask error.
